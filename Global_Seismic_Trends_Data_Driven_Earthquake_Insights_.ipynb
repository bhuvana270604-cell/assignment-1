```python
region_summary = df.groupby("country").agg(
    earthquake_frequency=("id", "count"),
    average_magnitude=("mag", "mean")
).reset_index()

region_summary = region_summary.sort_values(
    "earthquake_frequency",
    ascending=False
)

region_summary.head(10)
```

```python
# Normalize frequency and average magnitude
region_summary["frequency_score"] = (
    region_summary["earthquake_frequency"] /
    region_summary["earthquake_frequency"].max()
)

region_summary["magnitude_score"] = (
    region_summary["average_magnitude"] /
    region_summary["average_magnitude"].max()
)

# Combined seismic activity score
region_summary["seismic_activity_score"] = (
    region_summary["frequency_score"] * 0.5 +
    region_summary["magnitude_score"] * 0.5
)

# Rank regions
top_3_regions = region_summary.sort_values(
    "seismic_activity_score",
    ascending=False
).head(3)

top_3_regions[
    [
        "country",
        "earthquake_frequency",
        "average_magnitude",
        "seismic_activity_score"
    ]
]
```

# 1. Introduction

Earthquakes are natural phenomena that occur due to the sudden release of energy within the Earth's crust. Analyzing earthquake data helps us understand earthquake frequency, magnitude, depth, and geographical distribution.

This project, titled "DS Global Seismic Trends", analyzes earthquake data obtained from the USGS earthquake dataset. The analysis focuses on earthquake depth categories, geographical regions, earthquake frequency, and magnitude.

The project also identifies locations where both shallow and deep earthquakes occurred during the same month and determines the top three seismically active regions based on earthquake frequency and average magnitude.

# 2. Objectives

The main objectives of this project are:

1. To analyze global earthquake data based on depth, magnitude, frequency, and geographical location.

2. To categorize earthquakes into Shallow, Intermediate, and Deep based on their depth.

3. To identify countries or regions where both shallow and deep earthquakes occurred during the same month.

4. To identify the top three most seismically active regions by considering earthquake frequency and average magnitude.

5. To visualize earthquake patterns and derive meaningful insights from the dataset.

# 3. Dataset Description

The dataset used in this project is an earthquake dataset obtained from the USGS (United States Geological Survey).

The dataset contains 1,486 earthquake records and 22 original columns. The main columns used for this analysis include:

- time – Date and time when the earthquake occurred
- latitude – Geographic latitude of the earthquake
- longitude – Geographic longitude of the earthquake
- depth – Depth of the earthquake in kilometers
- mag – Magnitude of the earthquake
- magType – Magnitude calculation method
- place – Geographic location of the earthquake
- type – Type of seismic event

Additional columns were used during data cleaning, preprocessing, and analysis.

# 4. Data Cleaning and Preprocessing

The dataset was checked and prepared before performing the analysis.

The following preprocessing steps were performed:

1. The dataset was loaded into a Pandas DataFrame.

2. The dataset structure, columns, data types, and sample records were inspected.

3. Duplicate records were checked to ensure data quality.

4. Missing values were identified. The important columns used for the analysis, such as time, depth, magnitude, and place, contained no missing values.

5. The time column was converted into datetime format for time-based analysis.

6. Year and month were extracted from the time column.

7. Earthquakes were categorized into Shallow, Intermediate, and Deep based on their depth.

8. Location information was extracted from the place column for regional analysis.

After preprocessing, the dataset was ready for exploratory data analysis and visualization.

# 5. Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to understand the major patterns and characteristics of the earthquake dataset.

The analysis focused on:

- Distribution of earthquakes by depth category
- Earthquake frequency across countries and regions
- Locations where shallow and deep earthquakes occurred during the same month
- Identification of the most seismically active regions
- Relationship between earthquake depth and magnitude

Interactive visualizations were created using Plotly to make the earthquake patterns easier to understand and interpret.

## 5.1 Depth Distribution Analysis

The earthquakes were categorized into three depth groups: Shallow, Intermediate, and Deep.

The distribution of earthquakes was:

- Shallow: 1,291 earthquakes
- Intermediate: 176 earthquakes
- Deep: 19 earthquakes

The analysis shows that the majority of the recorded earthquakes in the dataset were Shallow earthquakes, while Deep earthquakes were comparatively fewer.

## 5.2 Top 10 Countries / Regions by Earthquake Frequency

The earthquake records were grouped by country or region to identify areas with the highest number of recorded earthquakes.

The top 10 countries or regions based on earthquake frequency were:

1. Russia – 1,020 earthquakes
2. Indonesia – 63 earthquakes
3. Japan – 28 earthquakes
4. Vanuatu – 23 earthquakes
5. Papua New Guinea – 22 earthquakes
6. Tonga – 20 earthquakes
7. Guam – 18 earthquakes
8. Philippines – 17 earthquakes
9. Argentina – 14 earthquakes
10. Drake Passage – 14 earthquakes

Russia had the highest number of recorded earthquake events in this dataset.

## 5.3 Same-Month Shallow and Deep Earthquake Analysis

The dataset was analyzed to identify countries or regions where both Shallow and Deep earthquakes occurred during the same month.

The analysis identified the following locations:

- Fiji – August 2025
- Indonesia – August 2025
- Japan – July 2025
- Kermadec Islands region – August 2025
- Papua New Guinea – August 2025
- Timor Leste – August 2025
- Tonga – August 2025
- South of the Fiji Islands – August 2025

A total of 8 country/region and month combinations were identified where both Shallow and Deep earthquakes occurred.

## 5.4 Top 3 Most Seismically Active Regions

The seismic activity of each country or region was evaluated using two factors:

- Earthquake frequency
- Average earthquake magnitude

A combined seismic activity measure was calculated by multiplying earthquake frequency by average magnitude.

Based on this measure, the top three most seismically active regions in the dataset were:

1. Russia – 1,020 earthquakes, average magnitude 4.81, combined activity 4,905.1
2. Indonesia – 63 earthquakes, average magnitude 4.78, combined activity 300.9
3. Japan – 28 earthquakes, average magnitude 4.86, combined activity 136.1

Russia had the highest combined seismic activity because it had a substantially higher number of recorded earthquake events in the dataset.

## 5.5 Magnitude vs Depth Analysis

A scatter plot was created to examine the relationship between earthquake depth and magnitude.

The visualization represents:

- X-axis: Earthquake Depth (km)
- Y-axis: Earthquake Magnitude
- Color categories: Shallow, Intermediate, and Deep

The visualization helps identify how earthquake magnitudes are distributed across different depth categories. It also provides a visual understanding of the relationship between earthquake depth and magnitude in the dataset.

# 6. Key Findings

The major findings from the earthquake data analysis are:

1. The dataset contains 1,486 earthquake records.

2. Shallow earthquakes were the most common, with 1,291 records, followed by Intermediate earthquakes with 176 records and Deep earthquakes with 19 records.

3. Eight country/region and month combinations were identified where both Shallow and Deep earthquakes occurred during the same month.

4. The identified locations included Fiji, Indonesia, Japan, Kermadec Islands region, Papua New Guinea, Timor Leste, Tonga, and south of the Fiji Islands.

5. Based on the combined seismic activity measure, Russia, Indonesia, and Japan were the top three regions in the dataset.

6. Russia recorded the highest earthquake frequency with 1,020 events.

7. The Magnitude vs Depth visualization showed how earthquake magnitudes were distributed across different depth categories.
